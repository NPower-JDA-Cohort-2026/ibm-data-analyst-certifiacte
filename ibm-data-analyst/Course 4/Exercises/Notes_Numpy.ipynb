**NumPy (Numerical Python)** is a powerful library for scientific computing in Python. It provides fast array operations, mathematical functions, and linear algebra tools.

**why use Numpy?**

Faster than Python lists (uses C under the hood)

Supports multi-dimensional arrays

Includes mathematical functions (e.g., sin, cos, mean, std)

Used in data science, machine learning, and AI

**Speed & Performance**

NumPy arrays are significantly faster than Python lists because:

They use compiled C code under the hood

Data is stored in contiguous memory blocks, which allows for vectorized operations

**Efficient Memory Usage**

NumPy uses less memory than regular Python lists since all elements in a NumPy array are of the same data type, which allows for tighter packing.

Vectorized Operations (No Loops)

You can perform operations on entire arrays without writing for loops. For example:

python

import numpy as np
a = np.array([1, 2, 3])
b = a * 2  # [2, 4, 6]


**Convenient Functions for Data Analysis & Math**

Includes built-in functions like mean(), sum(), std(), linspace(), etc.

Useful for statistics, linear algebra, and machine learning prep.

**Supports Multidimensional Arrays**

Easily handles 2D (matrices), 3D, or even higher-dimensional arrays.

**Foundation for Other Libraries**

Many data science libraries like Pandas, Scikit-learn, TensorFlow, and Matplotlib are built on top of NumPy.

```python
arr4=np.array([[2.5,3.3,4.4],[2.4,3.5,4.5],[4.2,4.3,4.4]])
print(arr4)
print(arr4.ndim)
print(arr4.shape)
print(arr4.size) #number of elements
print(arr4.itemsize)
print(arr4.dtype)
#It specifies the size in bytes
# of each element of the array. Data type int32 and
# float32 means each element of the array occupies
# 32 bits in memory. 8 bits form a byte. Thus, an
# array of elements of type int32 has itemsize 32/8=4
# bytes. Likewise, int64/float64 means each item
# has itemsize 64/8=8 bytes.
```
