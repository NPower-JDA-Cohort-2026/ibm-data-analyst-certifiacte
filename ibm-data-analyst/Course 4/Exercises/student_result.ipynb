```python
# Create students.txt without overwriting an existing file

try:
    file = open("students.txt", "x")
    file.close()
    print("students.txt created successfully.")
except FileExistsError:
    print("students.txt already exists. It was not overwritten.")
```

```python
# Save the first student's information
with open("students.txt", "w") as file:
    file.write("Alice, Python, 85\n")

# Keep asking for more students
while True:
    name = input("Enter student name (or 'q' to quit): ")

    if name.lower() == "q":
        break

    course = input("Enter course: ")
    grade = input("Enter grade: ")

    # Add new student on a new line
    with open("students.txt", "a") as file:
        file.write(f"{name}, {course}, {grade}\n")

print("Student information saved successfully!")
```

```python
# Add a new student to the existing file

with open("students.txt", "a") as file:
    file.write("Bob, Java, 90\n")

print("New student added successfully!")
```
