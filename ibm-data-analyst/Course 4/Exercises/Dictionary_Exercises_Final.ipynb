# Python Dictionary Exercises

```python
#Q1 Create a dictionary for all participants in the JDA program.
#Keys should be the first letter in their names,
#Values should be their names.
#[Hint] dic={"l":(Leul),"J":(Jerry,Jainam)........}
names={"A":("Asish","Ankita"),"G":"Ghunza","I":"Israt","J":"Jean","K":("Kumari","Kiran"),"P":("Prabesh","Passy")}
print(names)
```

```python
#Q2 What are the keys of your dictionary
print(f'The Keys of the dictionary are: {names.keys()}')
#To show it in the form of list as without the dict_keys object
print(f'The Keys of the dictionary are: {list(names.keys())}')
```

```python
#Q3 What are the values of your dictionary
print(f'The values of the dictionary are: {names.values()}')
```

```python
#Q4 Find participant's names that start with J,I,M,C,A
prefix_list = ('J','I','M','C','A')
for prefix in prefix_list:
  if prefix in names:
    print(f'The names starting with {prefix} are : {names.get(prefix)}')
```

```python
#Q5 Add the following to your dictionary
# Key<(K,PD)>, Value<(Kristina)>
names[("K","PD")] = "Kristina"
names
```

```python
#Q6 Find the length of your dictionary
print(f'The length of the dictionary is {len(names)}')
```

```python
#Q7 Add the following to your dictionary
# Key<(S,Instructor)>, Value<(Sarah)>
names[("S","Instructor")] = "Sarah"
names
```

```python
#Q8 Find the length of your dictionary
print(f'The length of the dictionary is {len(names)}')
```

```python
#Q7 Delete the following keys from your dictionary
#Key<(S,Instructor)>
#Key<J>
print(f'The dictionary before deletion is {names}')
#using the del keyword
if ("S","Instructor") in names:
  del names[("S","Instructor")]
#using the pop function, it also returns the deleted value
if "J" in names:
  names.pop("J")
print(f'The dictionary after deletion is {names}')
```

```python
#Q10 Create a new dictionary called synonyms_dic
synonyms_dic = {}
```

```python
#Q11 Add the synonyms of the following words to your dictionary
#Bad, Good, Awesome, Cold, Easy, Hard, Big, Small
synonyms_dic = {
    "Bad": "Awful",
    "Good": "Great",
    "Awesome":"Amazing",
    "Cold": "Chilly",
    "Easy": "Simple",
    "Hard": "Difficult",
    "Big": "Large",
    "Small": "Little"
}
print(synonyms_dic)
```

```python
#Q12 Use your dictionary to find the synonym of the following words
# Awesome, Easy, Small, Old
syn_list = ("Awesome","Easy","Small","Old")
for syn in syn_list:
  if syn in synonyms_dic:
    print(f'The synonym of the word {syn} is : {synonyms_dic[syn]}')
  else:
    print(f'The synonym of the word {syn} not found!')
```

```python
#Q17 Create the following dictionary
encryption= {10:"hungry",101:"am",110:"I",1011:",",1101:"eat",1001:"want",111:"to"}
decryption = "" #empty string
# What should be the decryption of the following code
decrpytion_code = "110 101 10 1011 110 1001 111 1101"
split_code = decrpytion_code.split()
for code in split_code:
  decryption += encryption[int(code)] + " "
print(decryption)
```
